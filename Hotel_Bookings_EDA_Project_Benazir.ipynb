# **Project Summary -**

In today’s fast-paced and highly competitive hospitality industry, companies like Booking.com handle vast volumes of reservation data across a diverse network of hotels, customer profiles, and booking channels. The ability to extract actionable insights from this data is crucial for improving customer satisfaction, reducing cancellations, optimizing room occupancy, and increasing overall profitability. This project, titled “Hotel Booking Analysis”, presents an in-depth exploratory data analysis (EDA) of a large hotel booking dataset, focusing on uncovering key trends, patterns, and relationships that influence booking behavior and outcomes.

The dataset includes detailed information about over 119,000 bookings made at two different hotel types—city hotels and resort hotels. It captures variables such as booking lead times, arrival dates, length of stay, type of meal plan, assigned room type, customer demographics, special requests, distribution channels, and whether or not the booking was eventually canceled. These features provide a rich foundation for understanding how different factors contribute to customer decisions and business performance.

The primary objective of this analysis is to explore the structure and content of the data, identify important variables, detect anomalies or inconsistencies, and uncover meaningful relationships between customer behavior and reservation outcomes. A particular emphasis is placed on booking cancellations, a critical issue in the hotel industry, as they directly impact revenue forecasting, resource allocation, and service efficiency.

The EDA is structured around several key questions:

What is the distribution of bookings across hotel types and seasons?

How does lead time affect the likelihood of cancellations?

What customer or booking characteristics are most associated with canceled reservations?

Are there any patterns in special requests, repeated guests, or market segments?

How do booking behaviors vary between city hotels and resort hotels?

Visualizations such as histograms, boxplots, count plots, pair plots, and heatmaps are used to investigate the data and highlight underlying trends. For example, a pairplot comparing variables like lead time, average daily rate (ADR), and number of special requests helps reveal correlations between guest expectations and cancellation behavior. Additional analysis of categorical variables like deposit type, distribution channel, and customer type sheds light on how booking channels and policies influence outcomes.

Key findings from the analysis include:

1. Cancellations are more frequent in city hotels compared to resort hotels.

2. Longer lead times are strongly associated with higher cancellation rates, suggesting uncertainty or changing travel plans.

3. No-deposit bookings are far more likely to be canceled than those requiring a deposit.

4. Guests who make more special requests or are repeat customers are significantly less likely to cancel, possibly reflecting more deliberate booking intent or stronger brand loyalty.

5. Certain distribution channels and market segments exhibit consistently higher cancellation rates, indicating potential areas for policy review or customer education.

These insights can help Booking.com and its partner hotels improve forecasting models, refine booking policies, and develop more personalized customer engagement strategies. For example, enforcing stricter deposit requirements for high-risk segments or offering incentives for repeat guests could help reduce cancellations and improve operational efficiency.

In conclusion, this project provides a data-driven foundation for understanding customer booking behavior in the hospitality sector. By leveraging exploratory data analysis techniques, it highlights the importance of data in shaping strategic decisions and improving service delivery in a dynamic, customer-centric environment.

After filling the null columns and deleting the duplicated rows, new outliers might show up. So now the dataset is checked for the presence of any new outliers. If outliers are present, they are to be removed.

I observed that the adr column may have outliers because the column mean (106.57) is much smaller than its maximum value which is 5400.00.

````python
# Checking outliers for adr column by plotting boxplot

sns.boxplot(data=data, x='adr') # Creating the boxplot for adr column

plt.title("Boxplot for detecting ouliers in adr column") # Setting the title for the boxplot

plt.show() # Displaying the boxplot created
````

The boxplot clearly shows the presence of outliers in adr column


````python
# Calculating the outliers thresholds and then filtering the data

q1=data['adr'].quantile(0.25) # Calculating the first quartile
q3=data['adr'].quantile(0.75) # Calculating the third quartile

iqr=q3-q1 # Calculating the inter quartile range

lower_fence=q1-(1.5*iqr) # Calculating the lower fence
upper_fence=q3+(1.5*iqr) # Calculating the upper fence

print(f"Lower fence: {lower_fence}")
print(f"Upper fence: {upper_fence}")
#Filtering the data of outliers
data=data[(data['adr']>=lower_fence) & (data['adr']<=upper_fence)]

# Finding the total rows after deletion of outliers
print(f"Total number of rows after deleting outliers: {len(data)}")
````

````python
# Defining correct month order
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

# Convert 'arrival_date_month' to categorical with the correct order
data["arrival_date_month"] = pd.Categorical(data["arrival_date_month"], categories=month_order, ordered=True)
````

````python
# Changing the datatypes of agent and company column to object since they are ID codes and not really numbers
# They are of no importance when it comes to performing calculations. Keeping them as number might cause analysis problems later on.
data['agent'] = data['agent'].astype('str')
data['company'] = data['company'].astype('str')
print(data['agent'].dtypes)
print(data['company'].dtypes)
````
